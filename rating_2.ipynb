```python
import csv
import random
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
STUDY_NR="s214696"
```

```python

folder = Path("synthetic")

images = sorted(p.name for p in folder.glob("*.png"))
trials = images * 10
random.shuffle(trials)

ratings = {name: [] for name in images}

for name in trials:
    plt.imshow(plt.imread(folder / name), cmap="gray")
    plt.axis("off")
    plt.show(block=False)
    plt.pause(0.1)

    r = input("Rating 1(male)-5(female) (q to quit): ")
    plt.close()

    if r == "q":
        break
    if r in "12345" and r:
        ratings[name].append(r)

with Path(f"experiment_2_{STUDY_NR}.csv").open("w", newline="", encoding="utf-8") as f:
    w = csv.writer(f)
    w.writerow(["filename"] + [f"rating_{i}" for i in range(1, 11)])
    for name in images:
        w.writerow([name] + ratings[name])

print("Done")
```
