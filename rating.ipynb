```python
import csv
from pathlib import Path
import random 
import matplotlib.pyplot as plt
```

```python

folder = Path("70+_clean")
csv_path=Path("photos.csv")

# read
with csv_path.open(newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

for row in rows:
    if row["rating"]:          # already rated -> skip
        continue

    img = plt.imread(folder / row["filename"])
    plt.imshow(img)
    plt.axis("off")
    plt.title(row["filename"])
    plt.show(block=False)
    plt.pause(0.1)

    rating = input("Rating 1(male)-5(female) (q to quit): ")
    plt.close()

    if rating == "q":
        break
    if rating in "12345" and rating:
        row["rating"] = rating

    # save after every picture so nothing is lost
    with csv_path.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=["filename", "rating"])
        writer.writeheader()
        writer.writerows(rows)

print("Done")
```
